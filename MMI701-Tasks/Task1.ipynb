# Preliminaries

```python
import numpy as np
```

# False Implementation

```python
def custom_transform(x):
  N = len(x)
  X_half = np.zeros(N//2 + 1, dtype=complex)
  for k in range(N//2 + 1 ):
    X_half[k] = np.sum(x * np.exp(-2j * np.pi * k * np.arange(N) / N))
  X_full = np.concatenate([X_half, np.conj(X_half[-2:0:-1])])
  return X_full
```

# Correct Implementation

```python
def custom_transform_correct(x):
    N = len(x)
    X = np.zeros(N, dtype=complex)
    for k in range(N):
        X[k] = np.sum(x * np.exp(-2j * np.pi * k * np.arange(N) / N)) # solve for all.
    return X
```

# Test

* A test case with random 1024-point as usual.
* first a only real valued random with zero complex then complex valued simple input.
* Use symmetry property using one and the corrected one and compare.
* In comparison it can be seen that other than some numerical errors, due to regular bit storing limit of python, no error is seen, so rounding made.

```python
N = 1024

# real and complex data generation
x_real = np.random.randn(N) + 0j*np.random.randn(N)
x_complex = np.random.randn(N) + 1j*np.random.randn(N)

# Transformation and inverse transformation (Reconstruction) and error check.
err_h_r = np.max(np.abs(x_real - np.fft.ifft(custom_transform(x_real))))
err_nh_r = np.max(np.abs(x_real - np.fft.ifft(custom_transform_correct(x_real))))
err_h_c = np.max(np.abs(x_complex - np.fft.ifft(custom_transform(x_complex))))
err_nh_c = np.max(np.abs(x_complex - np.fft.ifft(custom_transform_correct(x_complex))))

print("Err_max (real input) hermitian:", round(err_h_r,10), "non-hermitian:", round(err_nh_r,10))
print("Err_max (complex input) hermitian:", round(err_h_c,10) , "non-hermitian:", round(err_nh_c,10))
```
